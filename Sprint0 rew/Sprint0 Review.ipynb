# Спринт ревью #0 — ООП в C#

Проект для управления растениями в саду. Реализованы наследование и полиморфизм.

```
using System;

public class Plant
{
    public string Name { get; set; }
    public string Type { get; set; }
    public float Height { get; set; }

    public Plant(string name, string type, float height)
    {
        Name = name;
        Type = type;
        Height = height;
    }

    public virtual void Flowering()
    {
        Console.WriteLine($"Растение {Name} цветет.");
    }

    public void Grow(float growth)
    {
        Height += growth;
        Console.WriteLine($"{Name} выросло до {Height} см.");
    }
}

public class Flower : Plant
{
    public string Color { get; set; }
    public int PetalCount { get; set; }

    public Flower(string name, float height, string color, int petalCount)
        : base(name, "Цветок", height)
    {
        Color = color;
        PetalCount = petalCount;
    }

    public override void Flowering()
    {
        Console.WriteLine($"Цветок {Name} распускается и радует глаз!");
    }

    public void Wither()
    {
        Console.WriteLine($"Цветок {Name} начинает увядать.");
    }
}

public class Tree : Plant
{
    public int Age { get; set; }
    public string LeafType { get; set; }

    public Tree(string name, float height, int age, string leafType)
        : base(name, "Дерево", height)
    {
        Age = age;
        LeafType = leafType;
    }

    public override void Flowering()
    {
        Console.WriteLine($"Дерево {Name} покрывается новыми листьями!");
    }

    public void GrowTree(int years)
    {
        Age += years;
        Console.WriteLine($"Дереву {Name} теперь {Age} лет.");
    }
}
```

```
Flower rose = new Flower("Роза", 45.0f, "Красный", 20);
Tree oak = new Tree("Дуб", 250.0f, 10, "Лиственные");

rose.Flowering();
rose.Grow(5);
rose.Wither();

oak.Flowering();
oak.Grow(20);
oak.GrowTree(2);
```

## Демонстрация полиморфизма

Оба объекта хранятся в переменных типа `Plant`, но при вызове `Flowering()` выполняется переопределённая версия метода соответствующего класса.

```
Plant plant1 = new Flower("Тюльпан", 30.0f, "Желтый", 6);
Plant plant2 = new Tree("Береза", 300.0f, 15, "Лиственные");

plant1.Flowering();
plant2.Flowering();
```
